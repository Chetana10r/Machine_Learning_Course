Stemming

Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words known as lemma. Stemming is important in natural language understanding (NLU) and NLP.

eating, eat, eaten [eat]

```python
words=['eating','eaten','eat','writing','writes']
```

PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

RegexpStemmer Class

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('ingeating')
```

Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snow=SnowballStemmer('english')
```

```python
print(snow)
```

```python
for word in words:
    print(word+"---->"+snow.stem(word))
```
