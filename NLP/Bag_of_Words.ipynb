```python
import pandas as pd
from io import StringIO

messages = """ham\tWhat you doing?how are you?
ham\tOk lar... Joking wif u oni...
ham\t dun say so early hor... U c already then say...
ham\tMY NO. IN LUTON 0125698789 RING ME IF UR AROUND! H*
ham\tSiva is in hostel aha:-.
ham\tCos i was out shopping wif darren jus now n i called him 2 ask wat present he wan lor. Then he started guessing who i was wif n he finally guessed darren lor.
spam\tFreeMsg: Txt: CALL to No: 86888 & claim your reward of 3 hours talk time to use from your phone now! ubscribe6GBP/ mnth inc 3hrs 16 stop?txtStop
spam\tSunshine Quiz! Win a super Sony DVD recorder if you canname the capital of Australia? Text MQUIZ to 82277. B
spam\tURGENT! Your Mobile No 07808726822 was awarded a L2,000 Bonus Caller Prize on 02/09/03! This is our 2nd attempt to contact YOU! Call 0871-872-9758 BOX95QU"""

# Use StringIO to treat the multiline string as a file
message_file = StringIO(messages)

# Read from the file-like object
message_df = pd.read_csv(message_file, sep='\t', names=["label", "message"])

# Print the DataFrame to verify
print(message_df)
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    # 1. Remove non-alphabetic characters
    review = re.sub('[^a-zA-Z]', ' ', text)
    # 2. Convert to lowercase
    review = review.lower()
    # 3. Tokenize and remove stopwords
    review = review.split()
    review = [ps.stem(word) for word in review if word not in stop_words]
    # 4. Join the words back into a string
    return ' '.join(review)

# --- Apply the function to the 'message' column ---
corpus = message_df['message'].apply(preprocess_text).tolist()

print("Original Messages:")
print(message_df['message'])
print("\nProcessed Corpus:")
print(corpus)
```

Create Bag of WOrds

```python

## Create the Bag OF Words model
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True
cv=CountVectorizer(max_features=100,binary=True)
```

```python

X=cv.fit_transform(corpus).toarray()
```

```python

## Create the Bag OF Words model with ngram
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True
cv=CountVectorizer(max_features=100,binary=True,ngram_range=(2,3))
X=cv.fit_transform(corpus).toarray()
```
